```python
import math
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score


def foil_gain(p0, n0, p1, n1):
    if p1 == 0 or p1 + n1 == 0 or p0 == 0 or p0 + n0 == 0:
        return -float("inf")

    return p1 * (
        math.log2(p1 / (p1 + n1)) -
        math.log2(p0 / (p0 + n0))
    )


def covers(example, rule):
    for attribute, value in rule.items():
        if example[attribute] != value:
            return False
    return True


def foil(examples, target):
    Pos = [x for x in examples if x["Target"] == target]
    Neg = [x for x in examples if x["Target"] != target]

    learned_rules = []

    while Pos:
        NewRule = {}
        NewRuleNeg = Neg.copy()

        while NewRuleNeg:
            best_literal = None
            best_gain = -float("inf")

            attributes = [
                key for key in examples[0]
                if key != "Target" and key not in NewRule
            ]

            for attribute in attributes:
                values = set(x[attribute] for x in examples)

                for value in values:
                    p1 = sum(
                        1 for x in Pos
                        if x[attribute] == value
                    )

                    n1 = sum(
                        1 for x in NewRuleNeg
                        if x[attribute] == value
                    )

                    p0 = len(Pos)
                    n0 = len(NewRuleNeg)

                    gain = foil_gain(p0, n0, p1, n1)

                    if gain > best_gain:
                        best_gain = gain
                        best_literal = (attribute, value)

            if best_literal is None:
                break

            attribute, value = best_literal
            NewRule[attribute] = value

            NewRuleNeg = [
                x for x in NewRuleNeg
                if x[attribute] == value
            ]

        if NewRule:
            learned_rules.append(NewRule)

        Pos = [
            x for x in Pos
            if not covers(x, NewRule)
        ]

    return learned_rules


def classify(example, rules):
    for rule in rules:
        if covers(example, rule):
            return "Yes"

    return "No"


examples = [
    {"Sky": "Sunny", "Temp": "Warm", "Humidity": "Normal", "Target": "Yes"},
    {"Sky": "Sunny", "Temp": "Warm", "Humidity": "High", "Target": "No"},
    {"Sky": "Cloudy", "Temp": "Warm", "Humidity": "High", "Target": "Yes"},
    {"Sky": "Rainy", "Temp": "Cool", "Humidity": "High", "Target": "No"},
    {"Sky": "Sunny", "Temp": "Cool", "Humidity": "Normal", "Target": "Yes"},
    {"Sky": "Rainy", "Temp": "Warm", "Humidity": "Normal", "Target": "Yes"},
    {"Sky": "Rainy", "Temp": "Cool", "Humidity": "Normal", "Target": "No"}
]


rules = foil(examples, "Yes")


print("=" * 60)
print("                 FOIL ALGORITHM")
print("=" * 60)

print("\nLEARNED RULES")
print("-" * 60)

for i, rule in enumerate(rules, 1):
    conditions = [
        f"{attribute} = {value}"
        for attribute, value in rule.items()
    ]

    print(
        f"Rule {i}: IF "
        + " AND ".join(conditions)
        + " THEN Target = Yes"
    )


y_true = []
y_pred = []

for example in examples:
    actual = example["Target"]
    predicted = classify(example, rules)

    y_true.append(actual)
    y_pred.append(predicted)

    print(
        f"\nInput: Sky={example['Sky']}, "
        f"Temp={example['Temp']}, "
        f"Humidity={example['Humidity']}"
    )
    print(f"Actual Class    : {actual}")
    print(f"Predicted Class : {predicted}")


cm = confusion_matrix(
    y_true,
    y_pred,
    labels=["Yes", "No"]
)

print("\n" + "=" * 60)
print("              CONFUSION MATRIX")
print("=" * 60)

print("\n              Predicted")
print("             Yes     No")
print(f"Actual Yes   {cm[0][0]:<8}{cm[0][1]}")
print(f"Actual No    {cm[1][0]:<8}{cm[1][1]}")


accuracy = accuracy_score(y_true, y_pred)

print("\n" + "=" * 60)
print("                ACCURACY")
print("=" * 60)

print(f"Accuracy = {accuracy:.2f}")


print("\n" + "=" * 60)
print("           CLASSIFICATION REPORT")
print("=" * 60)

print(
    classification_report(
        y_true,
        y_pred,
        labels=["Yes", "No"],
        zero_division=0
    )
)

```
