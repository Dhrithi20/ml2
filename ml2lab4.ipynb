```python
import pandas as pd

data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
]

attributes = ['Sky', 'AirTemp', 'Humidity', 'Wind', 'Water', 'Forecast']
df = pd.DataFrame(data, columns=attributes + ['EnjoySport'])

S = ['Ø'] * len(attributes)
G = [['?'] * len(attributes)]

def consistent(h, e):
    return all(x == '?' or x == y for x, y in zip(h, e))

def more_general(h1, h2):
    return all(x == '?' or x == y for x, y in zip(h1, h2))

def generalize(s, e):
    s = s.copy()
    for i in range(len(s)):
        if s[i] == 'Ø':
            s[i] = e[i]
        elif s[i] != e[i]:
            s[i] = '?'
    return s

for _, row in df.iloc[:4].iterrows():
    example = row[attributes].tolist()
    target = row['EnjoySport']

    if target == 'Yes':
        G = [g for g in G if consistent(g, example)]
        if not consistent(S, example):
            S = generalize(S, example)
        G = [g for g in G if more_general(g, S)]

    else:
        if consistent(S, example):
            S = ['?'] * len(attributes)

        new_G = []
        for g in G:
            if consistent(g, example):
                for i in range(len(attributes)):
                    if g[i] == '?':
                        for value in df[attributes[i]].unique():
                            if value != example[i]:
                                new_g = g.copy()
                                new_g[i] = value
                                new_G.append(new_g)
            else:
                new_G.append(g)
        G = new_G

print("Specific Boundary:")
print(S)

print("General Boundary:")
for g in G:
    print(g)
```

```python
import pandas as pd

data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Warm', 'Same', 'Yes'],
    ['Rainy', 'Cold', 'High', 'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High', 'Strong', 'Cool', 'Change', 'Yes']
]

attributes = ['Sky', 'AirTemp', 'Humidity', 'Wind', 'Water', 'Forecast']
df = pd.DataFrame(data, columns=attributes + ['EnjoySport'])

S = ['Ø'] * 6
G = [['?'] * 6]

def covers(h, x):
    return all(a == '?' or a == b for a, b in zip(h, x))

def generalize(s, x):
    for i in range(len(s)):
        if s[i] == 'Ø':
            s[i] = x[i]
        elif s[i] != x[i]:
            s[i] = '?'
    return s

for n, row in df.iterrows():
    x = row[attributes].tolist()
    y = row['EnjoySport']

    if y == 'Yes':
        G = [g for g in G if covers(g, x)]

        if not covers(S, x):
            S = generalize(S, x)

        G = [g for g in G if covers(g, S)]

    else:
        new_G = []

        for g in G:
            if covers(g, x):
                for i in range(6):
                    if g[i] == '?':
                        values = df[attributes[i]].unique()

                        for value in values:
                            if value != x[i]:
                                new_g = g.copy()
                                new_g[i] = value

                                if covers(new_g, S):
                                    new_G.append(new_g)
            else:
                new_G.append(g)

        G = new_G

    print("\nStep", n + 1)
    print("S =", S)
    print("G =")
    for g in G:
        print(g)

print("\nSpecific Boundary:")
print(S)

print("\nGeneral Boundary:")
for g in G:
    print(g)
```
